# Download BDD100K Dataset

Downloads the BDD100K 2D detection subset to `/Volumes/T7/data-engine-dataset`.

**Before running:** Register for a free account at [bdd-data.berkeley.edu](https://bdd-data.berkeley.edu) and grab your download links. BDD100K requires authentication — there's no public direct URL.

We need two files:
- `bdd100k_images_100k.zip` (~6.8 GB) — the 100k keyframe images
- `bdd100k_labels_release.zip` (~130 MB) — detection annotations

```python
from pathlib import Path

DATA_ROOT = Path("/Volumes/T7/data-engine-dataset")
DOWNLOAD_DIR = DATA_ROOT / "downloads"
EXTRACT_DIR = DATA_ROOT / "bdd100k"

DOWNLOAD_DIR.mkdir(parents=True, exist_ok=True)
EXTRACT_DIR.mkdir(parents=True, exist_ok=True)

print(f"Download dir: {DOWNLOAD_DIR}")
print(f"Extract dir:  {EXTRACT_DIR}")
```

## Step 1: Paste your download URLs

Log into [bdd-data.berkeley.edu](https://bdd-data.berkeley.edu), navigate to the download page, and copy the URLs for:
1. **Images** — `bdd100k_images_100k.zip`
2. **Labels** — `bdd100k_labels_release.zip`

Paste them below. These are time-limited signed URLs.

```python
# Paste your signed download URLs here
IMAGES_URL = ""  # paste URL for bdd100k_images_100k.zip
LABELS_URL = ""  # paste URL for bdd100k_labels_release.zip

assert IMAGES_URL, "Paste your images download URL above"
assert LABELS_URL, "Paste your labels download URL above"
```

## Step 2: Download

```python
import subprocess

def download(url: str, dest_dir: Path, filename: str) -> Path:
    dest = dest_dir / filename
    if dest.exists():
        print(f"Already exists: {dest} ({dest.stat().st_size / 1e9:.1f} GB)")
        return dest
    print(f"Downloading {filename}...")
    subprocess.run(
        ["curl", "-L", "-o", str(dest), "--progress-bar", url],
        check=True,
    )
    print(f"Done: {dest} ({dest.stat().st_size / 1e9:.1f} GB)")
    return dest

images_zip = download(IMAGES_URL, DOWNLOAD_DIR, "bdd100k_images_100k.zip")
labels_zip = download(LABELS_URL, DOWNLOAD_DIR, "bdd100k_labels_release.zip")
```

## Step 3: Extract

```python
import zipfile

def extract(zip_path: Path, dest: Path) -> None:
    print(f"Extracting {zip_path.name} to {dest}...")
    with zipfile.ZipFile(zip_path, "r") as zf:
        zf.extractall(dest)
    print(f"Done.")

extract(images_zip, EXTRACT_DIR)
extract(labels_zip, EXTRACT_DIR)
```

## Step 4: Verify

```python
expected = {
    "train_images": EXTRACT_DIR / "bdd100k" / "images" / "100k" / "train",
    "val_images": EXTRACT_DIR / "bdd100k" / "images" / "100k" / "val",
    "labels": EXTRACT_DIR / "bdd100k" / "labels",
}

for name, path in expected.items():
    if path.exists():
        if path.is_dir():
            count = len(list(path.iterdir()))
            print(f"{name}: {path} ({count} items)")
        else:
            print(f"{name}: {path} (exists)")
    else:
        print(f"{name}: MISSING — {path}")

# Count train images — should be ~70,000
train_dir = expected["train_images"]
if train_dir.exists():
    n_train = len(list(train_dir.glob("*.jpg")))
    print(f"\nTrain images: {n_train:,} (expected ~70,000)")

val_dir = expected["val_images"]
if val_dir.exists():
    n_val = len(list(val_dir.glob("*.jpg")))
    print(f"Val images:   {n_val:,} (expected ~10,000)")
```

## Step 5: Symlink into project

Creates a symlink from the project's `data/raw` to the T7 drive so the pipeline code can find it.

```python
PROJECT_RAW = Path("/Users/Dylan/Documents/data_engine/data/raw")
LINK_TARGET = EXTRACT_DIR / "bdd100k"

symlink = PROJECT_RAW / "bdd100k"

if symlink.exists() or symlink.is_symlink():
    print(f"Symlink already exists: {symlink} -> {symlink.resolve()}")
else:
    symlink.symlink_to(LINK_TARGET)
    print(f"Created symlink: {symlink} -> {LINK_TARGET}")
```
